## Ejercicio 1: normalización, máscaras y binarización

1. Crea un array 2D de forma `(5, 5)` con números aleatorios entre 0 y 50.
2. Normalízalo al rango `[0, 1]` restando el mínimo y dividiendo entre el máximo menos el mínimo: `(valor - min) / (max - min)`.
3. Usa una máscara booleana para extraer únicamente los elementos mayores que `0.5`.
4. Binariza el array normalizado de dos formas: con una máscara booleana y con `np.where`. Los valores mayores que `0.5` pasan a `1`; los demás, a `0`.

Se fija una semilla para que los mismos valores aleatorios se obtengan al volver a ejecutar la celda. `integers(0, 51)` incluye el 0 y el 50. La normalización transforma el mínimo en 0 y el máximo en 1. Se comprueba también el caso excepcional en que todos los valores fueran iguales, porque entonces el denominador sería cero.

```python
generador = np.random.default_rng(42)
array = generador.integers(0, 51, size=(5, 5))

minimo = array.min()
maximo = array.max()

if maximo == minimo:
    normalizado = np.zeros(array.shape, dtype=float)
else:
    normalizado = (array - minimo) / (maximo - minimo)

mascara_mayores_que_05 = normalizado > 0.5
mayores_que_05 = normalizado[mascara_mayores_que_05]

binario_mascara = np.zeros_like(normalizado, dtype=int)
binario_mascara[mascara_mayores_que_05] = 1

binario_where = np.where(normalizado > 0.5, 1, 0)

print("Array original:\n", array)
print("\nMínimo y máximo:", minimo, maximo)
print("\nArray normalizado:\n", normalizado)
print("\nMáscara (True donde el valor es > 0.5):\n", mascara_mayores_que_05)
print("\nValores mayores que 0.5:\n", mayores_que_05)
print("\nBinarización con máscara:\n", binario_mascara)
print("\nBinarización con np.where:\n", binario_where)
```

La comparación `normalizado > 0.5` se aplica elemento a elemento y produce un array booleano con la misma forma `(5, 5)`. Al usarlo como índice, NumPy devuelve en un array unidimensional solo los valores donde la máscara es `True`. En la primera binarización, se crea una matriz de ceros y se asigna `1` en esas posiciones. `np.where(condicion, 1, 0)` hace la misma selección de forma directa; ambos resultados deben coincidir.

## Ejercicio 2: diagonal principal y secundaria

1. Crea una matriz de `5 × 5` con los valores del 0 al 24.
2. Extrae la diagonal principal y la diagonal secundaria.
3. Muestra cada resultado por separado.

`reshape(5, 5)` coloca los 25 valores en cinco filas y cinco columnas. `np.diag` extrae la diagonal principal. Para obtener la secundaria, `np.fliplr` invierte el orden de las columnas; la diagonal principal de esa matriz invertida corresponde a la diagonal secundaria original.

```python
matriz = np.arange(25).reshape(5, 5)
diagonal_principal = np.diag(matriz)
diagonal_secundaria = np.diag(np.fliplr(matriz))

print("Matriz:\n", matriz)
print("\nDiagonal principal:\n", diagonal_principal)
print("\nDiagonal secundaria:\n", diagonal_secundaria)
```

## Ejercicio 3: broadcasting con filas

1. Crea una matriz `6 × 4` de enteros aleatorios entre 0 y 20.
2. Calcula la media de cada fila.
3. Resta a cada elemento de la matriz la media de su fila usando broadcasting, sin bucles.
4. Explica las formas de los arrays y cómo se aplican las reglas de broadcasting.

`axis=1` indica que se calcula una media independiente a lo largo de las columnas de cada fila. Así se obtiene un valor por fila. La matriz tiene forma `(6, 4)` y el vector de medias forma `(6,)`. Para que cada media se reste a los cuatro elementos de su propia fila, se añade un eje con `[:, np.newaxis]`, convirtiendo el vector en una columna de forma `(6, 1)`.

NumPy compara las dimensiones desde la derecha: `4` y `1` son compatibles porque una dimensión de tamaño `1` se puede expandir a `4`. También, las seis filas coinciden. Por tanto, `(6, 4)` y `(6, 1)` se combinan y el resultado tiene forma `(6, 4)`. Esta expansión virtual permite operar sin bucles ni copiar explícitamente las medias cuatro veces.

```python
generador_filas = np.random.default_rng(123)
matriz_filas = generador_filas.integers(0, 21, size=(6, 4))

medias_filas = matriz_filas.mean(axis=1)
medias_columna = medias_filas[:, np.newaxis]
matriz_centrada = matriz_filas - medias_columna

print("Matriz original:\n", matriz_filas)
print("\nForma de la matriz:", matriz_filas.shape)
print("\nMedias de cada fila:\n", medias_filas)
print("Forma del vector de medias:", medias_filas.shape)
print("\nMedias como columna:\n", medias_columna)
print("Forma de la columna de medias:", medias_columna.shape)
print("\nMatriz con la media de cada fila restada:\n", matriz_centrada)
print("Forma del resultado:", matriz_centrada.shape)
print("\nMedia de cada fila del resultado (debe ser aproximadamente 0):\n", matriz_centrada.mean(axis=1))
```
